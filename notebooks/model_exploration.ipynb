# MNIST Model Exploration
Load and inspect MNIST, then view sample digits. Train the production model with `python train_model.py`.

```python
import matplotlib.pyplot as plt
import tensorflow as tf
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
print('Train:', x_train.shape, y_train.shape)
print('Test:', x_test.shape, y_test.shape)
```

```python
fig, axes = plt.subplots(2, 5, figsize=(10, 4))
for ax, image, label in zip(axes.flat, x_train[:10], y_train[:10]):
    ax.imshow(image, cmap='gray')
    ax.set_title(f'Label: {label}')
    ax.axis('off')
plt.tight_layout()
```

```python
x_train_scaled = x_train.astype('float32') / 255.0
x_test_scaled = x_test.astype('float32') / 255.0
print('Pixel range:', x_train_scaled.min(), 'to', x_train_scaled.max())
```
